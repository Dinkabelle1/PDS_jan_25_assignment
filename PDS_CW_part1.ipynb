## Q1.1: Reading in rows of a CSV file. [10 marks]

Write a function `csv_rows(filename, from_row=None, to_row=None)`
to read in the header and specific rows of a CSV file.
The input arguments of the function should be a string specifying a filename,
a starting row number
`from_row` and
a final row number `to_row`.
The row numbers should be either integers or None.
The starting row number can be 1 or higher or None.
Row 1 refers to the first row after the header.
If the starting row number is None, then the default value of 1 should be used.
The ending row number can be 1 or higher or None.
The ending row number refers to the last row to be included.
If the ending row number is None, then all rows starting at the starting row should be read in.
The function should return two outputs:
a list with the column names and list of rows, where each row is a list of values (strings or numbers).

You can use the file `part1.csv` for your tests.


A function was created to read a CSV file using basic Python (open() and readlines()). The header was extracted, and selected rows were obtained using a manual counter. Each row was cleaned, split into values, and numeric data was converted where possible. This demonstrates how CSV files can be processed without external libraries.

```python

def make_number(value):
    value = value.strip()
    try:
        return float(value)
    except:
        return value


def csv_rows(filename, from_row=None, to_row=None):
    if from_row is None:
        from_row = 1

    selected_rows = []

    with open(filename, "r") as file:
        lines = file.readlines()

    header = lines[0].strip().split(",")
    data_lines = lines[1:]

    row_number = 1

    for line in data_lines:
        if row_number >= from_row and (to_row is None or row_number <= to_row):
            row = line.strip().split(",")
            row = [make_number(value) for value in row]
            selected_rows.append(row)

        row_number += 1

    return header, selected_rows
```

```python
# Read rows 1 to 3 from the csv to test the code
header, rows = csv_rows("part1.csv", 1, 3)
print("Header:")
print(header)
print("\nFirst 3 rows:")
for row in rows:
    print(row)
```

## Q1.2: Reading in Columns of a CSV file. [10 marks]

Write a function `csv_cols(filename, from_row=None, to_row=None, columns=None)`
that takes as inputs a filename, a starting row, a final row and a list of column names.
The same conventions as for Q1.1 should be used for the starting and final row input parameters.
The column name parameter can either be a list of column names or None, which indicates all columns.
The function should return a dictionary with the selected column names as
keys and a list of values (strings or numbers) for the corresponding column.
Your function should use the function from Q1.1.

You can use the file `part1.csv` for your tests.



Another function was built to extract selected columns and store them in a dictionary format. The function from Q1.1 was reused to obtain the data, and column positions were identified using the header. 

```python
def csv_cols(filename, from_row=None, to_row=None, columns=None):
    header, rows = csv_rows(filename, from_row, to_row)

    if columns is None:
        columns = header

    data = {}

    for column in columns:
        column_index = header.index(column)
        data[column] = []

        for row in rows:
            data[column].append(row[column_index])

    return data
```

## Q1.3: Calculating the minimum, maximum and median of a list of values [10 marks]

Write a function `min_max_median(p)` to calculate
the minimum, maximum and median of a list of values.
The function should return the minimum, maximum and median.



A function was implemented to calculate the minimum, maximum, and median values from a dataset. Non-numeric values were filtered out to ensure accurate calculations. The data was sorted before computing the statistics, with both odd and even cases handled for the median. 

```python
def is_number(value):
    """Check if a value is an int or float."""
    return type(value) == int or type(value) == float


def min_max_median(p):
    """Calculate the minimum, maximum and median of a list of numeric values."""
    values = []

    for item in p:
        if is_number(item):
            values.append(item)

    if len(values) == 0:
        return None, None, None

    values = sorted(values)
    minimum = values[0]
    maximum = values[-1]

    middle = len(values) // 2

    if len(values) % 2 == 1:
        median = values[middle]
    else:
        median = (values[middle - 1] + values[middle]) / 2

    return minimum, maximum, median
```

```python
print(min_max_median([5, 2, 9, 1, 7]))       # odd number of values
print(min_max_median([5, 2, 9, 1]))          # even number of values
print(min_max_median([5, "-", 9, 1]))        # includes a missing value
```

Write a function `table_min_max_median(data, columns=None)` that takes as inputs
a dictionary with column names as keys and a list of values for each key.
The `columns` parameter can be either a list of column names or None, which
means all column names should be used.
The function should return a list of (minimum, maximum, median) tuples for each row,
and a list of (minimum, maximum, median) tuples for the selected columns.
Your function must use the function from Q1.3.

You can use the file `part1.csv` and the function from Q1.2 for your tests.

Another function was developed to calculate summary statistics for both rows and columns. Data was iterated across rows and columns using loops and passed into the statistical function created in Q1.3. 

```python
def table_min_max_median(data, columns=None):
    ## Calculate min, max and median for each row and for selected columns.
    if columns is None:
        columns = list(data.keys())

    row_stats = []
    col_stats = []

    # Number of rows is based on the first selected column.
    number_of_rows = len(data[columns[0]])

    # Row statistics
    for row_index in range(number_of_rows):
        row_values = []

        for column in columns:
            row_values.append(data[column][row_index])

        row_stats.append(min_max_median(row_values))
    # Column statistics
    for column in columns:
        col_stats.append(min_max_median(data[column]))

    return row_stats, col_stats
```

## Q1.5: Develop a Function to Print a Custom Table [10 marks]

Write a function `print_min_max_median(row_stats, col_stats, ...)`
that prints the results in a clear and attractive way.
The input parameters should be a list of (min, max, median) tuples for the rows
and a list of (min, max, median) tuples for the columns
(as generated by Q1.4).
You can add extra parameters for further customisation.

You can use the file `part1.csv` and the functions from Q1.2 and Q1.4 for your tests.

The function takes the results from Q1.4 and prints:
row number, minimum,  maximum, median.
It then prints the same type of information for the columns.
 

```python
def format_value(value):
    """Format numbers neatly for printing."""
    if value is None:
        return "None"
    return round(value, 2)


def print_min_max_median(row_stats, col_stats, column_names=None, max_rows=10):
    """Print row and column min, max and median statistics."""
    print("ROW STATISTICS")
    print("-" * 55)
    print("Row".ljust(10), "Minimum".ljust(15), "Maximum".ljust(15), "Median".ljust(15))
    print("-" * 55)

    for index, stats in enumerate(row_stats[:max_rows], start=1):
        minimum, maximum, median = stats
        print(
            str(index).ljust(10),
            str(format_value(minimum)).ljust(15),
            str(format_value(maximum)).ljust(15),
            str(format_value(median)).ljust(15)
        )

    if len(row_stats) > max_rows:
        print("... only the first", max_rows, "rows are shown")

    print("\nCOLUMN STATISTICS")
    print("-" * 70)
    print("Column".ljust(25), "Minimum".ljust(15), "Maximum".ljust(15), "Median".ljust(15))
    print("-" * 70)

    for index, stats in enumerate(col_stats):
        minimum, maximum, median = stats

        if column_names is None:
            name = "Column " + str(index + 1)
        else:
            name = column_names[index]

        print(
            name.ljust(25),
            str(format_value(minimum)).ljust(15),
            str(format_value(maximum)).ljust(15),
            str(format_value(median)).ljust(15)
        )
```
